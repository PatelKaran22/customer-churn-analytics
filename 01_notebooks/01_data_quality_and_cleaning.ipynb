```python


df = spark.table("workspace.default.raw_churn_customers")

print("Row count:", df.count())
display(df.limit(10))
```

```python

df.printSchema()
```

```python


from pyspark.sql.functions import col, sum as _sum

missing_values = df.select(
    [col(c).isNull().cast("int").alias(c) for c in df.columns]
).select(
    [_sum(c).alias(c) for c in df.columns]
)

display(missing_values)
```

```python


df_clean = df.drop("RowNumber", "Surname")

display(df_clean.limit(10))
```

```python
df_clean.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("workspace.default.clean_churn_customers")
```

```python


clean_df = spark.table("workspace.default.clean_churn_customers")

print(clean_df.count())
display(clean_df.limit(10))
```
